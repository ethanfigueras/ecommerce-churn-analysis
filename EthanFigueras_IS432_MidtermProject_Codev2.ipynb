```python
# Logistic Regression(Predicting Customer Churn)
from sklearn.linear_model import LogisticRegression

X = df[["Tenure", "WarehouseToHome", "HourSpendOnApp",
        "NumberOfDeviceRegistered", "SatisfactionScore",
        "Complain", "CouponUsed", "OrderCount",
        "DaySinceLastOrder", "CashbackAmount"]]
y = df["Churn"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
churn_prob = model.predict_proba(X_test)[:, 1]
X_test_results = X_test.copy()
X_test_results["PredictedProb_Churn"] = churn_prob

threshold = 0.5
X_test_results["PredictedChurn"] = X_test_results["PredictedProb_Churn"] >= threshold
X_test_results.head(10)
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy:.4f}")
```

```python
# K-Means Clustering
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
X = df[["Tenure", "CashbackAmount"]]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)
```

```python
colors = ['red', 'blue', 'green']
for cluster in range(3):
    cluster_data = df[df['Cluster'] == cluster]
    plt.scatter(cluster_data['Tenure'],
                cluster_data['CashbackAmount'],
                c=colors[cluster],
                label=f'Cluster {cluster}',
                alpha=0.6)

centroids = scaler.inverse_transform(kmeans.cluster_centers_)
plt.scatter(centroids[:, 0], centroids[:, 1],
            c='black', marker='X', s=200, label='Centroids')

plt.title("Customer Segments by Tenure and Cashback Amount")
plt.xlabel("Tenure")
plt.ylabel("Cashback Amount")
plt.legend()
plt.show()
```
